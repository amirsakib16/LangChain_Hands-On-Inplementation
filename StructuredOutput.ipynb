````python
import ollama
import json
import re
from typing import List
from pydantic import BaseModel

class Skill(BaseModel):
    name: str
    experience: int

class Student(BaseModel):
    name: str
    department: str
    skills: List[Skill]

prompt = """
Generate ONLY valid JSON.

Schema:
{
    "name": "string",
    "department": "string",
    "skills": [
        {
        "name": "string",
        "experience": integer
        }
    ]
}

Student Info:
Name: Saad
Department: CSE

Skills:
Python -> 4 years
Machine Learning -> 2 years
"""

response = ollama.chat(
    model="qwen2.5:1.5b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)

raw_output = response["message"]["content"]
print("=== RAW MODEL OUTPUT ===")
print(raw_output)

cleaned_output = re.sub(r"```json|```", "", raw_output).strip()
print("\n=== CLEANED JSON ===")
print(cleaned_output)
try:
    data = json.loads(cleaned_output)
    student = Student(**data)
    print("\n=== PYDANTIC OBJECT ===")
    print(student)
    print("\n=== TYPE ===")
    print(type(student))

except Exception as e:
    print("\nERROR:")
    print(e)
````
