```python
import ollama

def add(x: int, y: int) -> int:
    return x + y

tools = [
    {
        "type": "function",
        "function": {
            "name": "add",
            "description": "Add two numbers",
            "parameters": {
                "type": "object",
                "properties": {
                    "x": {
                        "type": "integer",
                        "description": "First number"
                    },
                    "y": {
                        "type": "integer",
                        "description": "Second number"
                    }
                },
                "required": ["x", "y"]
            }
        }
    }
]

messages = [
    {
        "role": "user",
        "content": "Add 100 and 250"
    }
]


response = ollama.chat(
    model="qwen2.5:1.5b",
    messages=messages,
    tools=tools
)

print("=== AI RESPONSE ===")
print(response["message"])


tool_calls = response["message"].get("tool_calls", [])

for tool_call in tool_calls:

    function_name = tool_call["function"]["name"]
    arguments = tool_call["function"]["arguments"]

    if function_name == "add":

        result = add(
            x=arguments["x"],
            y=arguments["y"]
        )

        print("\n=== TOOL RESULT ===")
        print(result)

        # assistant tool call
        messages.append(response["message"])

        # tool output
        messages.append({
            "role": "tool",
            "content": str(result)
        })


final_response = ollama.chat(
    model="qwen2.5:1.5b",
    messages=messages
)

print("\n=== FINAL RESPONSE ===")
print(final_response["message"]["content"])
```

```python
import ollama

def add(x: int, y: int) -> int:
    """This function adds two numbers. Usefull to calculate total/accumulate/cummulative sum."""
    return x + y

def subtract(x: int, y: int) -> int:
    """This function subtracts two numbers. Usefull to calculate the decrease amount or the difference between two numbers."""
    return x - y

def multiply(x: int, y: int, z: int) -> int:
    """This function multiplies three numbers. Usefull to calculate the total quantity when you have multiple items of the same type."""
    return x * y * z

def divide(x: int, y: int) -> float:
    """This function divides two numbers. Usefull to calculate the average or the ratio between two quantities, times of one number than another."""
    return x / y

available_functions = {
    "add": add,
    "subtract": subtract,
    "multiply": multiply,
    "divide": divide
}

tools = [
    {
        "type": "function",
        "function": {
            "name": "add",
            "description": """This function adds two numbers. Usefull to calculate total/accumulate/cummulative sum.""",
            "parameters": {
                "type": "object",
                "properties": {
                    "x": {"type": "integer"},
                    "y": {"type": "integer"}
                },
                "required": ["x", "y"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "subtract",
            "description": """This function subtracts two numbers. Usefull to calculate the decrease amount or the difference between two numbers.""",
            "parameters": {
                "type": "object",
                "properties": {
                    "x": {"type": "integer"},
                    "y": {"type": "integer"}
                },
                "required": ["x", "y"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "multiply",
            "description": """This function multiplies three numbers. Usefull to calculate the total quantity when you have multiple items of the same type.""",
            "parameters": {
                "type": "object",
                "properties": {
                    "x": {"type": "integer"},
                    "y": {"type": "integer"},
                    "z": {"type": "integer"}
                },
                "required": ["x", "y", "z"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "divide",
            "description": """This function divides two numbers. Usefull to calculate the average or the ratio between two quantities, times of one number than another.""",
            "parameters": {
                "type": "object",
                "properties": {
                    "x": {"type": "integer"},
                    "y": {"type": "integer"}
                },
                "required": ["x", "y"]
            }
        }
    }
]

messages = [
    {
        "role": "user",
        "content": "A person has 10 banana, 20 mangoes. how many total banana and mangoes do he have? What happen if I donate 6 bananas to my friend, what will be the total amount of banana and mangoes I have? and Lastly,How many times of mangos than banana I have after donating 6 bananas to my friend?"
    }
]


response = ollama.chat(
    model="qwen2.5:1.5b",
    messages=messages,
    tools=tools
)

print("=== AI RESPONSE ===")
print(response["message"])

tool_calls = response["message"].get("tool_calls", [])

for tool_call in tool_calls:

    function_name = tool_call["function"]["name"]
    arguments = tool_call["function"]["arguments"]

    print("\n=== TOOL CALLED ===")
    print(function_name)

    print("\n=== ARGUMENTS ===")
    print(arguments)

    function_to_call = available_functions.get(function_name)
    if function_to_call:
        result = function_to_call(**arguments)
        print("\n=== TOOL RESULT ===")
        print(result)
        messages.append(response["message"])
        messages.append({
            "role": "tool",
            "content": str(result)
        })

final_response = ollama.chat(
    model="qwen2.5:1.5b",
    messages=messages)
print("\n=== FINAL RESPONSE ===")
print(final_response["message"]["content"])
```
