```python
import ollama

messages = [
    {
        "role": "system",
        "content": "You are a cybersecurity expert."
    },

    {
        "role": "user",
        "content": "What is SQL Injection?"
    },

    {
        "role": "assistant",
        "content": "SQL Injection is a database attack where malicious SQL queries are inserted into input fields."
    },

    {
        "role": "user",
        "content": "Can it bypass login systems?"
    }
]

response = ollama.chat(
    model="qwen2.5:1.5b",
    messages=messages
)

print(response["message"]["content"])
```

```python
import ollama

messages = [
    {
        "role": "system",
        "content": "You are a Computer Security expert."
    },

    {
        "role": "user",
        "content": "What is a virus?"
    },

    {
        "role": "assistant",
        "content": "A virus is a type of malicious entity."
    },

    {
        "role": "user",
        "content": "Is it harmful for humans?"
    },

    {
        "role": "system",
        "content": "Suppose you are a doctor."
    },

    {
        "role": "user",
        "content": "What is a virus?"
    },

    {
        "role": "assistant",
        "content": "A virus is a type of malicious entity."
    },
    {
        "role": "user",
        "content": "Is it harmful for humans?"
    },
]

response = ollama.chat(
    model="qwen2.5:1.5b",
    messages=messages
)

print(response["message"]["content"])
```
