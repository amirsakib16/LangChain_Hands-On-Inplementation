```python
import ollama
prompts = ["What is Deep Learning?", "What is Computer Vision?", "What is NLP?"]

for i in prompts:
    print(f"\n\nPrompt: {i}\nAnswer: ", end="", flush=True)

    stream = ollama.chat(
        model="qwen2.5:1.5b",
        messages=[{"role": "user", "content": i}],
        stream=True
    )

    for chunk in stream:
        print(chunk["message"]["content"], end="", flush=True)
```

```python
import ollama
from concurrent.futures import ThreadPoolExecutor

prompts = ["What is Deep Learning?", "What is Computer Vision?", "What is NLP?"]

def stream_ask(prompt):
    stream = ollama.chat(
        model="qwen2.5:1.5b",
        messages=[{"role": "user", "content": prompt}],
        stream=True
    )

    result = ""
    for chunk in stream:
        result += chunk["message"]["content"]
    return result

with ThreadPoolExecutor() as executor:
    responses = list(executor.map(stream_ask, prompts))

# print(responses)
for i, r in enumerate(responses):
    print(f"\nResponse {i+1}:\n{r}")
```
